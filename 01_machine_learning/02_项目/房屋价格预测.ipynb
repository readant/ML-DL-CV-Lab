# 房屋价格线性回归案例

核心公式：
$$ y = w \cdot x + b $$
- $y$：模型预测输出（预测房价）
- $w$：权重（斜率，coef_）
- $x$：输入特征（房子面积）
- $b$：偏置/截距（intercept_）

```python
# 简单的线性回归示例(调包最简版)
import numpy as np
# sklearn 传统机器学习库，LinearRegression 是现成的线性回归模型
from sklearn.linear_model import LinearRegression
```

## 变量
1. `house_sizes`:
   - 含义：**特征矩阵 X**，模型的输入
   - 内容：`[[50],[70],[90],[110]]`,房子面积
   - 形状：`(样本数量, 特征数量)` ———>`(4,1)`
      - 4 = 4条样本（4套房子）
      - 1 = 每个样本只有1个特征：面积
    - 为什么要`reshape(-1,1)`
      - `np.array([50,70,90,110])`是一维`[50,70,90,110]` ,shape=(4,)
      - sklearn 硬性规则：**特征X必须是二位数组**，不能一维
      - `-1`:自动计算这一维的长度，这里自动算出4;`1`:每个样本1个特征
    - 注意事项：
      - 预测的时候输入也要保持一样的格式：`[[120]]`,不能直接写`[120]`
      - 如果以后是多个特征（面积 + 楼层），就变成`(4,2)`
2. `prices`:
   - 含义：**标签y/真实值**，模型要去学习拟合的目标
   - 内容：`[300,400,500,600]`,对应房价
   - 形状：一维`(4,)`,标签y可以是一维，sklearn允许
   - 注意事项： 
     - X和y**样本数量必须对齐**：这里X 4个样本，y 也必须4个，多一个少一个直接报错
     - 标签是**连续数字**，这叫【回归任务】；如果是类别(0,1)，那就是分类任务
3. `model`:
   - 含义：模型实例对象，训练好的模型，可以理解成一个空白的“学习容器”
   - 内容：`model = LinearRegression()` 这一步：只是创建容器，还没学习任何数据，`w、b都是空的`
   - 注意事项:
     - 没执行`.fit()`之前不要去读`model.coef_`，会直接报错，参数还没算出来

```python
# 房子面积数据
house_sizes = np.array([50,70,90,110]).reshape(-1,1)
# 对应房价（万元）
prices = np.array([300,400,500,600])
```

- `model = LinearRegression()`:创建一个空的线性回归模型
- `.fit(X,y)`:训练（拟合）,也就是让机器从这4组数据里面算出一条直线\(y=wx+b\)

```python
# 创建模型并训练
model = LinearRegression()
model.fit(house_sizes,prices)
# 预测120平房子的价格
prediction= model.predict([[120]])
print(f"预测房价:{prediction[0]:.1f}万元")
```
