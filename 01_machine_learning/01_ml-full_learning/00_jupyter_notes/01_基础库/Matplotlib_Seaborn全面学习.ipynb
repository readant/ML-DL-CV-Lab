# Matplotlib + Seaborn 数据可视化

Matplotlib 是 Python 最基础的绑图库，Seaborn 基于 Matplotlib 提供更美观的统计图表。二者配合使用是数据分析和 ML 模型评估的标配。

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# 中文显示设置（Windows）
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
# 统一图表风格
sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 100

print(f"Matplotlib: {plt.matplotlib.__version__}")
print(f"Seaborn: {sns.__version__}")
````

---
## 1. Matplotlib 基础

### 1.1 Figure 与 Axes 的关系

```
Figure（画布）
 └── Axes（坐标系，一个 Figure 可含多个 Axes）
      ├── axis（x轴、y轴）
      └── plot（实际绑图）
```

````python
# 方式1: plt 快捷方式（简单场景）
x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('三角函数')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
````

````python
# 方式2: 面向对象方式（推荐，更灵活）
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), color='blue', linewidth=2, linestyle='-', label='sin')
ax.plot(x, np.cos(x), color='red', linewidth=2, linestyle='--', label='cos')
ax.set_title('三角函数（面向对象）', fontsize=14)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 2. 常用图表类型

### 2.1 折线图（趋势展示）

````python
np.random.seed(42)
days = pd.date_range('2024-01-01', periods=30)
temperature = np.random.normal(25, 5, 30)
humidity = np.random.normal(60, 10, 30)

fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(days, temperature, 'o-', color='tomato', label='温度(°C)')
ax1.set_ylabel('温度(°C)', color='tomato')
ax1.tick_params(axis='y', labelcolor='tomato')

# 双Y轴
ax2 = ax1.twinx()
ax2.plot(days, humidity, 's--', color='steelblue', label='湿度(%)')
ax2.set_ylabel('湿度(%)', color='steelblue')
ax2.tick_params(axis='y', labelcolor='steelblue')

ax1.set_title('温湿度变化趋势')
fig.legend(loc='upper right', bbox_to_anchor=(0.9, 0.88))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

### 2.2 柱状图（分类对比）

````python
categories = ['手机', '电脑', '耳机', '平板', '手表']
sales_q1 = [120, 80, 200, 90, 150]
sales_q2 = [150, 95, 180, 110, 170]

x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))
bars1 = ax.bar(x - width/2, sales_q1, width, label='Q1', color='steelblue')
bars2 = ax.bar(x + width/2, sales_q2, width, label='Q2', color='coral')

# 在柱子上方标数值
for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
            f'{int(bar.get_height())}', ha='center', fontsize=9)
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
            f'{int(bar.get_height())}', ha='center', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_title('季度销售额对比')
ax.legend()
plt.tight_layout()
plt.show()
````

### 2.3 散点图（相关性分析）

````python
np.random.seed(42)
n = 100
study_hours = np.random.uniform(1, 10, n)
scores = 40 + 5 * study_hours + np.random.normal(0, 5, n)
sizes = np.random.uniform(50, 200, n)

fig, ax = plt.subplots(figsize=(8, 6))
scatter = ax.scatter(study_hours, scores, c=sizes, cmap='viridis',
                     alpha=0.7, s=sizes*0.5, edgecolors='white')
plt.colorbar(scatter, label='参与度')

# 拟合线
z = np.polyfit(study_hours, scores, 1)
p = np.poly1d(z)
x_line = np.linspace(1, 10, 100)
ax.plot(x_line, p(x_line), 'r--', alpha=0.8, label=f'拟合线 y={z[0]:.1f}x+{z[1]:.1f}')

ax.set_xlabel('学习时间(小时)')
ax.set_ylabel('考试成绩')
ax.set_title('学习时间 vs 考试成绩')
ax.legend()
plt.tight_layout()
plt.show()
````

### 2.4 直方图 + KDE（分布展示）

````python
np.random.seed(42)
class_a = np.random.normal(75, 10, 200)
class_b = np.random.normal(68, 15, 200)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 直方图
axes[0].hist(class_a, bins=20, alpha=0.6, label='A班', color='steelblue')
axes[0].hist(class_b, bins=20, alpha=0.6, label='B班', color='coral')
axes[0].set_title('成绩分布（直方图）')
axes[0].set_xlabel('分数')
axes[0].set_ylabel('频次')
axes[0].legend()

# KDE（核密度估计）
from scipy.stats import gaussian_kde
x_range = np.linspace(30, 110, 200)
kde_a = gaussian_kde(class_a)
kde_b = gaussian_kde(class_b)
axes[1].fill_between(x_range, kde_a(x_range), alpha=0.4, label='A班', color='steelblue')
axes[1].fill_between(x_range, kde_b(x_range), alpha=0.4, label='B班', color='coral')
axes[1].set_title('成绩分布（KDE）')
axes[1].set_xlabel('分数')
axes[1].set_ylabel('密度')
axes[1].legend()

plt.tight_layout()
plt.show()
````

### 2.5 箱线图（数据离散度）

````python
np.random.seed(42)
subjects = ['数学', '英语', '物理', '化学', '生物']
data = [np.random.normal(70 + i*3, 10 + i*2, 100) for i in range(5)]

fig, ax = plt.subplots(figsize=(8, 5))
bp = ax.boxplot(data, labels=subjects, patch_artist=True,
                boxprops=dict(facecolor='lightblue'))

# 给每个箱体不同颜色
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#FF99CC']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

ax.set_title('各科成绩分布')
ax.set_ylabel('分数')
plt.tight_layout()
plt.show()
````

### 2.6 饼图（占比）

````python
labels = ['手机', '电脑', '耳机', '平板', '手表']
sizes = [35, 25, 20, 12, 8]
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7']
explode = (0.05, 0, 0, 0, 0)  # 突出第一块

fig, ax = plt.subplots(figsize=(7, 7))
wedges, texts, autotexts = ax.pie(
    sizes, explode=explode, labels=labels, colors=colors,
    autopct='%1.1f%%', shadow=True, startangle=90,
    textprops={'fontsize': 12}
)
for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_fontweight('bold')

ax.set_title('产品销售占比')
plt.tight_layout()
plt.show()
````

## 3. 子图布局

````python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# 折线图
axes[0, 0].plot(np.linspace(0, 10, 50), np.sin(np.linspace(0, 10, 50)))
axes[0, 0].set_title('折线图')

# 柱状图
axes[0, 1].bar(['A', 'B', 'C', 'D'], [3, 7, 2, 5], color='skyblue')
axes[0, 1].set_title('柱状图')

# 散点图
axes[0, 2].scatter(np.random.rand(50), np.random.rand(50), c=np.random.rand(50), cmap='plasma')
axes[0, 2].set_title('散点图')

# 直方图
axes[1, 0].hist(np.random.randn(200), bins=20, color='salmon', edgecolor='white')
axes[1, 0].set_title('直方图')

# 饼图
axes[1, 1].pie([30, 25, 20, 15], labels=['A', 'B', 'C', 'D'], autopct='%1.1f%%')
axes[1, 1].set_title('饼图')

# 箱线图
axes[1, 2].boxplot([np.random.randn(100) + i for i in range(4)])
axes[1, 2].set_title('箱线图')

plt.suptitle('Matplotlib 常用图表一览', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
````

## 4. Seaborn 统计可视化

Seaborn 优势：自动统计计算、内置主题、与 Pandas DataFrame 无缝配合

````python
# 准备示例数据
np.random.seed(42)
n = 200
df = pd.DataFrame({
    '数学': np.random.normal(75, 12, n),
    '英语': np.random.normal(70, 15, n),
    '物理': np.random.normal(68, 18, n),
    '性别': np.random.choice(['男', '女'], n),
    '班级': np.random.choice(['A班', 'B班', 'C班'], n),
    '学习时间': np.random.uniform(1, 8, n)
})
df['总分'] = df['数学'] + df['英语'] + df['物理']
````

````python
# 4.1 散点图 + 回归线
fig, ax = plt.subplots(figsize=(8, 5))
sns.regplot(data=df, x='学习时间', y='总分', ax=ax,
            scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})
ax.set_title('学习时间 vs 总分（含回归线）')
plt.tight_layout()
plt.show()
````

````python
# 4.2 分类散点图
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.stripplot(data=df, x='班级', y='数学', hue='性别', ax=axes[0], alpha=0.6)
axes[0].set_title('各班数学成绩分布（stripplot）')

sns.swarmplot(data=df.sample(50), x='班级', y='数学', hue='性别', ax=axes[1], size=4)
axes[1].set_title('各班数学成绩（swarmplot）')

plt.tight_layout()
plt.show()
````

````python
# 4.3 小提琴图
fig, ax = plt.subplots(figsize=(8, 5))
sns.violinplot(data=df, x='班级', y='数学', hue='性别', split=True, ax=ax)
ax.set_title('各班数学成绩（小提琴图）')
plt.tight_layout()
plt.show()
````

````python
# 4.4 热力图（相关系数矩阵）
fig, ax = plt.subplots(figsize=(7, 6))
corr = df[['数学', '英语', '物理', '学习时间', '总分']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlBu_r',
            center=0, square=True, ax=ax, linewidths=0.5)
ax.set_title('特征相关系数热力图')
plt.tight_layout()
plt.show()
````

````python
# 4.5 pairplot（多变量关系矩阵）
sns.pairplot(df[['数学', '英语', '物理', '班级']], hue='班级',
             diag_kind='kde', height=2.5)
plt.suptitle('多变量关系 pairplot', y=1.02)
plt.show()
````

````python
# 4.6 分布图
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

sns.histplot(data=df, x='数学', kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('数学成绩分布')

sns.kdeplot(data=df, x='数学', hue='班级', fill=True, ax=axes[1])
axes[1].set_title('各班数学 KDE')

sns.ecdfplot(data=df, x='数学', hue='班级', ax=axes[2])
axes[2].set_title('数学成绩 ECDF')

plt.tight_layout()
plt.show()
````

## 5. 图表美化技巧

````python
# 预设主题
themes = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
fig, axes = plt.subplots(1, 5, figsize=(18, 3))

for ax, theme in zip(axes, themes):
    with sns.axes_style(theme):
        sns.lineplot(x=np.linspace(0, 10, 50), y=np.sin(np.linspace(0, 10, 50)), ax=ax)
        ax.set_title(f'{theme}')

plt.tight_layout()
plt.show()
````

````python
# 调色板
fig, axes = plt.subplots(1, 3, figsize=(14, 3))

sns.barplot(x=['A', 'B', 'C', 'D', 'E'], y=[5, 3, 7, 2, 8],
            palette='pastel', ax=axes[0])
axes[0].set_title('pastel')

sns.barplot(x=['A', 'B', 'C', 'D', 'E'], y=[5, 3, 7, 2, 8],
            palette='muted', ax=axes[1])
axes[1].set_title('muted')

sns.barplot(x=['A', 'B', 'C', 'D', 'E'], y=[5, 3, 7, 2, 8],
            palette='deep', ax=axes[2])
axes[2].set_title('deep')

plt.tight_layout()
plt.show()
````

## 总结

### Matplotlib 核心
| 图表 | 用途 |
|------|------|
| 折线图 `plot` | 趋势、时序 |
| 柱状图 `bar` | 分类对比 |
| 散点图 `scatter` | 相关性分析 |
| 直方图 `hist` | 分布展示 |
| 箱线图 `boxplot` | 离散度、异常值 |
| 饼图 `pie` | 占比 |
| 子图 `subplots` | 多图布局 |

### Seaborn 核心
| 图表 | 用途 |
|------|------|
| `regplot` | 散点 + 回归线 |
| `heatmap` | 相关系数矩阵 |
| `pairplot` | 多变量关系矩阵 |
| `violinplot` | 分布形态对比 |
| `histplot/kdeplot` | 密度分布 |
